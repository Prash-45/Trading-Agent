```python
import pandas as pd

from trading_env import TradingEnv
```

```python
stocks = ["AAPL", "AMZN", "JNJ", "JPM", "MSFT", "TSLA"]

dataframes = []

for stock in stocks:
    df = pd.read_csv(f"data/train/{stock}.csv")
    df["Date"] = pd.to_datetime(df["Date"])
    dataframes.append(df)
```

```python
env = TradingEnv(dataframes)
```

```python
obs, info = env.reset()

print(obs.shape)
print(obs[-3:])
```

```python
print(f"Number of stocks loaded: {len(dataframes)}")

for df in dataframes:
    print(df["Ticker"].iloc[0], df.shape)
```

```python
from stable_baselines3 import A2C

model = A2C(
    policy="MlpPolicy",
    env=env,
    learning_rate=3e-4,
    gamma=0.99,
    n_steps=5,
    ent_coef=0.01,
    verbose=1,
    seed=42
)
```

```python
model.learn(total_timesteps=300000)
```

```python
model.save("models/a2c_reward_v3")
```
