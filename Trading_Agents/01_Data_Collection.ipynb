```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import yfinance as yf
```

```python
tickers = [
    "AAPL",
    "MSFT",
    "AMZN",
    "JPM",
    "TSLA",
    "JNJ"
]

df = yf.download(
    tickers=tickers,
    start="2015-01-01",
    end="2025-12-31",
    interval="1d",
    auto_adjust=True,
    group_by="ticker",
    progress=True
)
```

```python
stock_data = []

for ticker in tickers:
    temp = df[ticker].copy()
    temp.reset_index(inplace=True)
    temp["Ticker"] = ticker
    stock_data.append(temp)

stock_df = pd.concat(stock_data, ignore_index=True)

stock_df.head()
```

```python
stock_df.to_csv("stocks_cleaned.csv", index=False)
```
